```python
from __future__ import division
#practical example 2 matrix multiplication  tensorflow

import tensorflow as tf
#create tensor
tensor_a = tf.constant([[1, 2], [3, 4]])
tensor_b = tf.constant([[5, 6], [7, 8]])

# Addition
addition = tf.add(tensor_a, tensor_b)
print("Addition (A + B):",
      addition)

# Subtraction
Subtraction = tf.subtract(tensor_a, tensor_b)
print("Subtraction (A - B):",
      Subtraction)

# Multiplication (element-wise)
multiplication = tf.multiply(tensor_a, tensor_b)
print("Element-wise Multiplication (A * B):",
      multiplication)

# Division (element-wise)
division = tf.divide(tensor_a, tensor_b)
print("Element-wise Division (A / B):",
      division)
```

#building neural network
building neural network is the process of creating the structure of ann by adding layers,neurons and activation functions using tensorflow and keras.
#step to build a neural netowrk

import libraries -> creating sequeantial model-> add input layer-> add hidden layers -> add ouput layers -> compile model -> train model-> make predictions

#components of neural networks

1. input layer
2. hidden layer
3. ouput layer


```python
from tensorflow import keras
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense

# CREATE NEURAL NETWORK
model = Sequential()

# Define the input shape using an Input layer
model.add(Input(shape=(4,)))

# FIRST DENSE LAYER
model.add(Dense(
    units=16,
    activation="relu"
))

# HIDDEN LAYER
model.add(Dense(
    units=8,
    activation="relu"
))

# hidden layer
model.add(Dense(
    units=4,
    activation="relu"
))

# OUTPUT LAYER
model.add(Dense(
    units=1,
    activation="sigmoid"
))


#display model summary
model.summary()
```

#model compilation in tensorflow $ keras
model compilation is the process of configuring a neural network by specifing the optimizer loss function and evaluation metrices before training

components
1. optimizer - an optimizer is adjust of the neural network to minimize the loss and improve prediction accuracy
#adam.
#sgd
#rmsrop
#adgrad
#2. loss function the loss function calculates error of the neural network
#binary_crossentropy
#categorial_crossentropy
#mean_squared_error(mse)
3. metrics - metrices evaluate how well the neural network is pending or performing.


```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import Input

# Create model
model = Sequential()

# Use Input layer for shape
model.add(Input(shape=(4,)))

# Add layers
model.add(Dense(16, activation="relu"))
model.add(Dense(8, activation="relu"))
model.add(Dense(4, activation="relu"))
model.add(Dense(1, activation="sigmoid"))

# Compile model (corrected from 'models' to 'model')
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Display summary
model.summary()
```

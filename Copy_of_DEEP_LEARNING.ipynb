```python
from tensorflow import keras
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense

# CREATE NEURAL NETWORK
model = Sequential()

# Define the input shape using an Input layer
model.add(Input(shape=(4,)))

# FIRST DENSE LAYER
model.add(Dense(
    units=16,
    activation="relu"
))

# HIDDEN LAYER
model.add(Dense(
    units=8,
    activation="relu"
))

# hidden layer
model.add(Dense(
    units=4,
    activation="relu"
))

# OUTPUT LAYER
model.add(Dense(
    units=1,
    activation="sigmoid"
))


#display model summary
model.summary()
```

Backpropagation in Artificial Neural Nettwork(ANN)

Backpropagation is the process of sending the prediction error back  throught the neural network to update weight and  improve accurancy

WHY DO WE NEED BACKPROPAAGATION

1.make a predicition

2.calculate the error loss

3.send the error backward

4.update the weight

5.improve the next preediction

WORKFLOW

input data--> forward propagation--> prediction-->calculates loss-->backpropagation-->update weight-->repeart for next  epoach

```python
# ==========================================================
# BACKPROPAGATION EXAMPLE USING TENSORFLOW & KERAS
# ==========================================================

# Import Required Libraries
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import pandas as pd

# ==========================================================
# STEP 1 : Load Dataset
# ==========================================================

df = pd.read_csv("/content/train.csv")

# ==========================================================
# STEP 2 : Data Preprocessing
# ==========================================================

# Select Features
X = df[["Pclass","Sex","Age","SibSp","Parch","Fare","Embarked"]]

# Target Variable
y = df["Survived"]

# Fill Missing Values
X["Age"] = X["Age"].fillna(X["Age"].mean())

# Encode Categorical Columns
X["Sex"] = X["Sex"].map({"male":0,"female":1})
X["Embarked"] = X["Embarked"].map({"S":0,"C":1,"Q":2})

X = X.fillna(0)

# Feature Scaling
sampler = StandardScaler()
X = sampler.fit_transform(X)

# ==========================================================
# STEP 3 : Split Dataset
# ==========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# ==========================================================
# STEP 4 : Build Neural Network
# ==========================================================

model = Sequential([
    tf.keras.Input(shape=(X_train.shape[1],)), # Explicit Input layer
    Dense(units=16, activation="relu"),
    Dense(units=8, activation="relu"),
    Dense(units=1, activation="sigmoid")
])

# ==========================================================
# STEP 5 : Compile Model
# ==========================================================

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# ==========================================================
# STEP 6 : Train Model
# ==========================================================

history = model.fit(
    X_train,
    y_train,
    validation_split=0.20,
    epochs=200,
    batch_size=32,
    verbose=1
)

# ==========================================================
# STEP 7 : Evaluate Model
# ==========================================================

loss, accuracy = model.evaluate(X_test, y_test)

print("Loss :", loss)
print("Accuracy :", accuracy)
```

Hyperparameter tuning

Hyperparameter tuning is the process of hinding  the optimal value of hyperparameter to improve a model's accurancy and reduce error

COMMON HYPERPARAMETER

1.learning rate

2.epochs

3.batch size

4.optimizer

5.hidden layer

6.number of neural

7.activation function

8.dropout rate



```python
# ==========================================================
# HYPERPARAMETER TUNING USING TENSORFLOW & KERAS
# ==========================================================

# Import Required Libraries
import pandas as pd
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

# ==========================================================
# STEP 1 : Load Dataset
# ==========================================================

df = pd.read_csv("/content/train.csv")

# ==========================================================
# STEP 2 : Data Preprocessing
# ==========================================================

# Select Features
X = df[["Pclass","Sex","Age","SibSp","Parch","Fare","Embarked"]]

# Target Variable
y = df["Survived"]

# Fill Missing Values
X["Age"] = X["Age"].fillna(X["Age"].mean())

# Convert Categorical Columns into Numbers
X["Sex"] = X["Sex"].map({"male":0,"female":1})
X["Embarked"] = X["Embarked"].map({"S":0,"C":1,"Q":2})

# Fill Remaining Missing Values
X = X.fillna(0)

# Feature Scaling
scaler = StandardScaler()
X = scaler.fit_transform(X)

# ==========================================================
# STEP 3 : Train-Test Split
# ==========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# ==========================================================
# STEP 4 : Hyperparameters
# ==========================================================

learning_rates = [0.1, 0.01, 0.001]

batch_sizes = [16, 32]

epochs_list = [20, 50]

# Variables to Store Best Result
best_accuracy = 0
best_parameters = {}

# ==========================================================
# STEP 5 : Hyperparameter Tuning
# ==========================================================

for lr in learning_rates:

    for batch in batch_sizes:

        for epochs in epochs_list:

            print("="*60)
            print("Learning Rate :", lr)
            print("Batch Size    :", batch)
            print("Epochs        :", epochs)

            # Build Model
            model = Sequential()

            model.add(Dense(
                16,
                activation="relu",
                input_shape=(X_train.shape[1],)
            ))

            model.add(Dense(
                8,
                activation="relu"
            ))

            model.add(Dense(
                1,
                activation="sigmoid"
            ))

            # Create Optimizer
            optimizer = Adam(learning_rate=lr)

            # Compile Model
            model.compile(
                optimizer=optimizer,
                loss="binary_crossentropy",
                metrics=["accuracy"]
            )

            # Train Model
            history = model.fit(
                X_train,
                y_train,
                validation_split=0.20,
                epochs=epochs,
                batch_size=batch,
                verbose=0
            )

            # Get Highest Validation Accuracy
            accuracy = max(history.history["val_accuracy"])

            print("Validation Accuracy :", accuracy)

            # Store Best Hyperparameters
            if accuracy > best_accuracy:

                best_accuracy = accuracy

                best_parameters = {

                    "Learning Rate": lr,

                    "Batch Size": batch,

                    "Epochs": epochs
                }

# ==========================================================
# STEP 6 : Display Best Hyperparameters
# ==========================================================

print("\n")
print("="*60)
print("BEST HYPERPARAMETERS")
print("="*60)

print("Best Validation Accuracy :", best_accuracy)

print(best_parameters)
```
